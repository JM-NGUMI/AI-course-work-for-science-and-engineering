```python
def celsius_to_fahrenheit(celsius):
    """Converts Celsius to Fahrenheit."""
    return (celsius * 9/5) + 32

def main():
    threshold_f = 100.0

    print("--- Temperature Converter ---")
    print(f"Current threshold: {threshold_f}°F\n")

    while True:
        user_input = input("Enter a temperature in °C (or type 'quit' to exit): ").strip()

        if user_input.lower() == 'quit':
            print("Exiting program. Goodbye!")
            break

        try:
            celsius = float(user_input)

            fahrenheit = celsius_to_fahrenheit(celsius)

            print(f"{celsius:.1f}°C is equal to {fahrenheit:.1f}°F")

            if fahrenheit > threshold_f:
                print(f"Status: Above the threshold of {threshold_f}°F.\n")
            elif fahrenheit < threshold_f:
                print(f"Status: Below the threshold of {threshold_f}°F.\n")
            else:
                print(f"Status: Exactly at the threshold of {threshold_f}°F.\n")

        except ValueError:
            print("Invalid input. Please enter a valid numerical temperature.\n")

if __name__ == "__main__":
    main()
```

# How the Program work

# The program work by receiving a temperature value from the user in degree celsius, first convert it to Farenheight and then monitors it with a set value in Farenheight and give out a tatus information whether higher or lower than the set valu.The program daoes this repeatedly until quit command is inputed in place of temperature value
